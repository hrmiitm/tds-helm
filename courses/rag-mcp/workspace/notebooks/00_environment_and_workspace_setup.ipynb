### Step 2: Inspect the Corporate Policy Dataset
Our retrieval system will search across corporate policy documents stored in `data/corporate_policies/`.
Let's inspect the files and view the first few lines of each policy.

```python
data_dir = Path("data/corporate_policies")
if not data_dir.exists():
    data_dir = Path("/workspace/data/corporate_policies")

policies = sorted(list(data_dir.glob("*.txt")))
print(f"Found {len(policies)} policy documents in {data_dir}:\n")

for policy_file in policies:
    content = policy_file.read_text(encoding="utf-8")
    preview = content.split("\n")[:4]
    print(f"=== {policy_file.name} ({len(content)} characters) ===")
    print("\n".join(preview))
    print("-" * 50)
```

### Step 3: Verify the Offline SentenceTransformer Model
To perform semantic search, we use dense embeddings generated by the `all-MiniLM-L6-v2` model.
This model is pre-cached and baked directly inside the environment for completely offline execution.
Let's load the model and generate a test embedding for a sample sentence.

```python
import os
import numpy as np
from sentence_transformers import SentenceTransformer

# Check for baked offline model path
model_path = os.environ.get("BAKED_MODEL_DIR", "/models/all-MiniLM-L6-v2")
if not os.path.exists(model_path):
    model_path = "all-MiniLM-L6-v2"

print(f"Loading embedding model from: {model_path}...")
model = SentenceTransformer(model_path)
print("[OK] Model loaded successfully!")

# Generate a test embedding
sample_text = "Employees may work remotely up to two days per week with manager approval."
embedding = model.encode(sample_text)

print(f"\nSample text: '{sample_text}'")
print(f"Embedding dimension: {embedding.shape[0]} (vector length)")
print(f"L2 Norm of embedding: {np.linalg.norm(embedding):.4f} (unit hypersphere vector)")
print(f"First 5 vector values: {embedding[:5].round(4)}")
```
